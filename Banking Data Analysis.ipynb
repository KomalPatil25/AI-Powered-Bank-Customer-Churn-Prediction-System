# Outliers

```python
import matplotlib.pyplot as plt
import seaborn as sns

def checkOutliers(data, col):
    plt.figure(figsize=(12,4))
    
    # Boxplot
    plt.subplot(1,2,1)
    plt.boxplot(data[col])
    plt.title(f"Boxplot of {col}")
    
    # Histogram 
    plt.subplot(1,2,2)
    sns.histplot(data=data, x=col, kde=True)
    
    mean = data[col].mean()
    std = data[col].std()
    
    plt.axvline(mean + 3*std, color='red')
    plt.axvline(mean - 3*std, color='red')
    
    plt.tight_layout()
    plt.show()
```

```python
checkOutliers(data, 'Monthly_Income')
```

```python
checkOutliers(data, 'Loan_Amount')
```

```python
checkOutliers(data, 'Credit_Score')
```

```python
checkOutliers(data, 'Card_Spend')
```

```python
checkOutliers(data, 'Customer_Tenure') 
```

# Handle Outliers

```python
def handleOuteliers(data, col):
    Q3 = data[col].quantile(0.75)
    Q1 = data[col].quantile(0.25)
    
    IQR = Q3 - Q1
    
    upper = Q3 + (1.5 * IQR)
    lower = Q1 - (1.5 * IQR)
    
    data.loc[data[col] > upper, col] = upper
    data.loc[data[col] < lower, col] = lower
```

```python
handleOuteliers(data, 'Credit_Score')
```

```python
checkOutliers(data, 'Credit_Score')
```

# Data Transformation

### Create Income Category

```python
def income_category(x):
    if x < 50000:
        return 'Low'
    elif x < 100000:
        return 'Medium'
    else:
        return 'High'


data['Income_Category'] = data['Monthly_Income'].apply(income_category)
```

### Create Age Group

```python
bins = [18,30,45,60,100]
labels = ['Young','Adult','Middle Age','Senior']


data['Age_Group'] = pd.cut(data['Age'], bins=bins, labels=labels)
```

### Create Loan Risk Category

```python
def risk(score):
    if score >= 750:
        return 'Low Risk'
    elif score >= 600:
        return 'Medium Risk'
    else:
        return 'High Risk'


data['Risk_Category'] = data['Credit_Score'].apply(risk)
```

# Run SQL Queries in Jupyter

### View Data

```python
query = "SELECT * FROM bank_customers LIMIT 5"

pd.read_sql(query, conn)
```

### Average Income by City

```python
query = """
        SELECT City,
               AVG(Monthly_Income) AS avg_income
        FROM bank_customers
        GROUP BY City
        ORDER BY avg_income DESC
        """

pd.read_sql(query, conn)
```

### Total Customers by Gender

```python
query = """
SELECT Gender, COUNT(*) as total_customers
FROM bank_customers
GROUP BY GENDER
"""
pd.read_sql(query, conn)
```

### High Risk Customers

```python
query = """
SELECT Customer_ID, Credit_Score, Loan_Amount 
from bank_customers
WHERE Credit_Score < 500
"""
pd.read_sql(query, conn)
```

### Top 10 Customers by Income

```python
query = """
SELECT Customer_ID,
       Monthly_Income
FROM bank_customers
ORDER BY Monthly_Income DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```
